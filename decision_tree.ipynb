```python
encoding = {"mesin" : {"bensin" : 0, "diesel" : 1},
            "penggerak" : {"depan" : 0, "belakang" : 1}}
# Load the CSV file into a pandas DataFrame named 'data'
data = pd.read_csv('decisiontree_1.csv')
# Apply the encoding using the .replace() method
data.replace(encoding, inplace=True)
```

```python
data.head()
```

```python
x = data.drop(['ID', 'label'], axis=1)
y = data['label']
```

```python
import sklearn.model_selection as ms
X_train, X_test, y_train, y_test = ms.train_test_split(x,
                                                       y,
                                                       test_size=0.2)
```

```python
import sklearn.tree as tree
model = tree.DecisionTreeClassifier(criterion='gini', max_depth=5)
model.fit(X_train, y_train)
```

```python
y_prediksi = model.predict(X_test)
y_prediksi
```

```python
import sklearn.metrics as met
print (met.accuracy_score(y_test, y_prediksi))
```

```python
import matplotlib.pyplot as plt
from sklearn import tree

plt.subplots(figsize=(10,10))
tree.plot_tree(model, fontsize=10)
plt.show()
```
